```python
from syncode import Syncode

import warnings
warnings.filterwarnings('ignore')

model_name = "microsoft/phi-2"

# Load the unconstrained original model
llm = Syncode(model=model_name, mode='original', max_new_tokens=300)

# Load the Syncode augmented model
syn_llm = Syncode(model = model_name, grammar='./python.lark', parse_output_only=False, indent=True, max_new_tokens=300)
```

```python
prefix = "import unittest\n\nclass testCases(unittest.TestCase):"
output = llm.infer(prefix)[0]
print(f"LLM output:\n{prefix+output}\n")
```

```python
output = syn_llm.infer(prefix)[0]
print(f"Syncode Output: \n{prefix+output}")
```
